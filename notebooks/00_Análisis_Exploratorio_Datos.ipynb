### Análisis Exploratorio de Datos

En primer lugar, necesitamos conocer los conjuntos de datos con los que vamos a trabajar. Conocer las columnas de cada dataframe, saber qué indica el dato en esa columna y crear un diccionario de datos es fundamental para poder comenzar cualquier análisis.

En nuestro caso hemos descargado los archivos GTFS de la EMT de la ciudad de Madrid desde el Portal de Datos Abiertos de la EMT en la siguiente url: https://datos.emtmadrid.es/dataset/gtfs-de-emtmadrid.

En esos datos contamos con los archivos "routes.txt", "stops.txt", "calendar_dates.txt", "calendar.txt", "frequencies.txt", "shapes.txt", "stop_times.txt" y "trips.txt".

Procedemos a realizar un análisis exploratorio de los datos descargados, conociendo los nombres de columna de cada archivo, la estructura y volumen de los mismos, comprobando duplicados y valores faltantes o nulos,  Queremos conocer también el tamaño de cada conjunto de datos y si hay duplicados o valores faltantes o nulos.

Para ello usaremos Python y las librerías os, pandas, numpy y Path.

```python
# Imports mínimos
import os
import pandas as pd
import numpy as np
from pathlib import Path

# Rutas
project_root = Path.cwd().parent
RAW = project_root / "data" / "raw"
GTFS_DIR = RAW / "gtfs"
AGENCY_GTFS = GTFS_DIR / "agency.txt"
CALENDAR_DATES_GTFS = GTFS_DIR / "calendar_dates.txt"
CALENDAR = GTFS_DIR / "calendar.txt"
FREQUENCIES_GTFS = GTFS_DIR / "frequencies.txt"
ROUTES_GTFS = GTFS_DIR / "routes.txt"
SHAPES_GTFS = GTFS_DIR / "shapes.txt"
STOP_TIMES_GTFS = GTFS_DIR / "stop_times.txt"
STOPS_GTFS = GTFS_DIR / "stops.txt"
TRIPS_GTFS = GTFS_DIR / "trips.txt"
OUT_SUMMARY = project_root / "data" / "processed" / "schema_summary_ingest.json"

# Carga de datos y almacenamiento en memoria global
agency = pd.read_csv(AGENCY_GTFS)
agency.to_parquet('agency_copy.parquet')


frequencies = pd.read_csv(FREQUENCIES_GTFS)
frequencies.to_parquet('frequencies_copy.parquet')


stop_times = pd.read_csv(STOP_TIMES_GTFS)
stop_times.to_parquet('stop_times_copy.parquet')
```
